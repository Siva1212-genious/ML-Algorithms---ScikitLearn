```python
#Now lets remove unusefull columns
raw_data.drop(columns = ['Unnamed: 32'], inplace = True)
```

```python
#Now lets do the scalling for all the input columns using MinMaxScalar
```

```python
#Lets first choose the numeric columns
import numpy as np

numeric_columns = raw_data.select_dtypes(include = np.number).columns.tolist()
```

```python
numeric_columns
```

```python
numeric_columns = numeric_columns[1:]
```

```python
numeric_columns
```

```python
#Now Letus do the encoding for the 'object columns' i.e,. diagnosis

diagnosis_codes = {'B': 0, 'M': 1}

raw_data['diagnosis_codes'] = raw_data.diagnosis.map(diagnosis_codes)
```

```python
cancer_no = raw_data[raw_data['diagnosis_codes'] == 0]
```

```python
raw_data
```

```python
#Now lets use the scalling for numeric columns
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

raw_data[numeric_columns] = scaler.fit_transform(raw_data[numeric_columns])
```

```python
raw_data
```

```python
raw_data.describe()
```

```python
#Now first lets use Logestic regression model and after that we will use Decision Trees and Random Forest.

#we will split the train_data, test_data, validation_data
```

```python
main_data = raw_data[numeric_columns]
```

```python
main_data.info()
```

```python
target_data = raw_data['diagnosis_codes']
```

```python
from sklearn.model_selection import train_test_split

train_val_data, test_data, train_val_target, test_target = train_test_split(raw_data, target_data, test_size = 0.2, random_state = 42)

train_data, val_data, train_target, val_target = train_test_split(train_val_data,train_val_target, test_size = 0.2, random_state = 42)
```

```python
print(train_data.shape)
print(val_data.shape)
print(test_data.shape)
print(train_target.shape)
```

```python
numeric_columns
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(train_data[numeric_columns], train_target)
```

```python
train_prediction = model.predict(train_data[numeric_columns])
```

```python
train_prediction
```

```python
#Now Lets calculate the train probability for train_data, val_data
```

```python
model.classes_
```

```python
import numpy as np

prediction_prob = model.predict_proba(train_data[numeric_columns])

np.set_printoptions(suppress = True)

print(prediction_prob)
```

```python
#Now we need to check the accuracy of the predicted and actual data for training data
```

```python
from sklearn.metrics import accuracy_score

train_data_accuracy = accuracy_score(train_target, train_prediction, normalize=True)

print("Accuracy: ", train_data_accuracy*100)
```

```python
#So, our training data accuracy is 97.25%
#Now we will check the confusion matrix for this training data.
```

```python
############################## IMPORTANT ###################################################################
#We Need to use CONFUSION MATRIX to identify, where our model is actually getting failed.
#This CONFUSION MATRIX defines, where our model is actually getting failed.
#It gives us 1.TRUE NEGATIVE(TN), 2. TRUE POSITIVE(TP), 3. FALSE NEGATIVE(FN), 4. FALSE POSITIVE(FP)
# With This Confusion matrix we can extract 1.ACCURACY 2. PRECISION 3. RECALL 4.F1-SCORE
#1.ACCURACY = (TN + TP)/(TN + TP+ FN+ FP) #
#2. PRECISION = TP/(TP + FP) # Of everything the model predicted positive, how many were actually positive?
#3. RECALL = TP/(TP + FN) # Of everything that was actually positive, how many did the model find?
#4. F1-SCORE = F1 = 2 × (Precision × Recall) / (Precision + Recall) # Combines precision and recall:
```

```python
#NOW lets calculate the CONFUSION MATRIX for the training data.
```

```python
from sklearn.metrics import confusion_matrix

confusion_mat = confusion_matrix(train_target, train_prediction)

print(confusion_mat)
```

```python
############################################### IMPORTANT ###################################################
# Now we can find TP, FP, TN, FN from this confusion matrix and we can find accuracy, precision, recall, f1-score
```

```python
TN, FP, FN, TP = confusion_matrix(train_target, train_prediction).ravel()

print("True Negative:", TN)
print("False Positive:", FP)
print("False Negative:", FN)
print("True Positive:", TP)
```

```python
#lETS TEST BY GIVING ONE SINGLE ROW
import pandas as pd

def predict_random_data(single_row):
  input_data = pd.DataFrame([single_row])
  input_data[numeric_columns] = scaler.transform(input_data[numeric_columns])
  X_input = input_data[numeric_columns]

  new_prediction = model.predict(X_input)[0]
  new_pred_prob = model.predict_proba(X_input)[0][list(model.classes_).index(new_prediction)]


  return new_prediction, new_pred_prob
```
