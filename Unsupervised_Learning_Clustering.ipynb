```python
import matplotlib.pyplot as plt
import seaborn as sns

iris_data = sns.load_dataset('iris')

iris_data
```

```python
sns.scatterplot(data=iris_data, x='petal_length', y='petal_width', hue='species')
plt.show()
```

```python
from sklearn.cluster import KMeans
import numpy as np

x = iris_data.select_dtypes(include=np.number).columns.tolist()

x
```

```python
kmean = KMeans(n_clusters = 3, random_state=42)

predicted_values = kmean.fit_predict(iris_data[x])

predicted_values
```

```python
iris_data['Predicted_value'] = predicted_values

iris_data
```

```python
sns.scatterplot(data=iris_data, x='petal_length', y='petal_width', hue='Predicted_value')

plt.show()

sns.scatterplot(data=iris_data, x='petal_length', y='petal_width', hue='species')
```

```python
import pandas as pd

data_set = pd.read_csv('./customer-segmentation-tutorial-in-python/Mall_Customers.csv')
data_set_1 = pd.read_csv('./customer-segmentation-tutorial-in-python/Mall_Customers.csv')

data_set
```

```python
# According to my understanding, we need to classify our customers based on age, annual income, spending_score.
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

numeric_cols = data_set.select_dtypes(include=np.number).columns.tolist()

scalar_cols = numeric_cols[1:]

scalar_cols
```

```python
scaler = StandardScaler()

data_set[scalar_cols] = scaler.fit_transform(data_set[scalar_cols])

data_set
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

range_ = range(1,11)
inertias=[]

for cluster in range_:
  k__mean = KMeans(n_clusters=cluster, max_iter=250)

  k__mean.fit_predict(data_set[scalar_cols])

  inertias.append(k__mean.inertia_)
  print(cluster, k__mean.inertia_)

sns.lineplot(x=range_, y=inertias, marker='O')
plt.show()
```

```python
k___mean = KMeans(n_clusters = 4, max_iter=250)

predictions = k___mean.fit_predict(data_set[scalar_cols])

predictions
```

```python
data_set['Clusters'] = predictions
#Now i will join this for non_scalarized datset also
data_set_1['Clusters'] = predictions
```

```python
data_set_1
```

```python
from seaborn.widgets import color_palette
sns.scatterplot(data=data_set_1, x='Age', y='Spending Score (1-100)', hue='Clusters', palette="Set2")
plt.show()
```
