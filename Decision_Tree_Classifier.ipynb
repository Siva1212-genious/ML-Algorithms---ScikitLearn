```python

import opendatasets as od

dataset_url = 'https://www.kaggle.com/jsphyg/weather-dataset-rattle-package'

od.download(dataset_url)
```

```python
import os

data_dir = './weather-dataset-rattle-package'

os.listdir(data_dir)
```

```python
dataset = data_dir + '/weatherAUS.csv'
```

```python
import pandas as pd

raw_data = pd.read_csv(dataset)

raw_data
```

```python
# Now, we will clean the data by seperating the numerical_columns and categorical_columns,
# Numerical_Columns will be cleaned by imputing technique and then we can either scale or we can just leave them like that.
# Categorical_Columns can be cleared by filling the empty columns with 'unknown' and we can encode them.
```

```python
import numpy as np

numeric_colm = raw_data.select_dtypes(include = np.number).columns.tolist()
categorical_colm = raw_data.select_dtypes(include = 'object').columns.tolist()
```

```python
numeric_colm, categorical_colm
```

```python
# Let's clean the numeric columns first
```

```python
raw_data[numeric_colm].isna().sum()
```

```python
# Now we will use imputer function to clean the empty columns in by using mean strategy
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy = 'mean')

raw_data[numeric_colm] = imputer.fit(raw_data[numeric_colm]).transform(raw_data[numeric_colm])
```

```python
raw_data[numeric_colm].isna().sum()
```

```python
# Our numeric columns are good with no empty columns.

# Now we can focus on the categorical columns and replace empty columns with 'unknown'
```

```python
raw_data[categorical_colm] = raw_data[categorical_colm].fillna('unknown')
```

```python
raw_data[categorical_colm].isna().sum()
```

```python
# now we need to drop all the columns which have unknown in 'RainToday', 'RainTomorrow'. Because,
```

```python
#filtered_data = raw_data[raw_data['RainToday'] & raw_data['RainTomorrow'] != 'unknown']   #Not Working

filtered_data = raw_data[(raw_data['RainToday'] != 'unknown') & (raw_data['RainTomorrow'] != 'unknown')]
```

```python
filtered_data[filtered_data['RainToday'] == 'unknown']
```

```python
filtered_data[filtered_data['RainTomorrow'] == 'unknown']
```

```python
# Now we can remove the 'Date', 'RainTomorrow' columns
```

```python
input_columns = list(filtered_data)[1:-1]
target_columns = 'RainTomorrow'
```

```python
input_columns, target_columns
```

```python
Main_data = filtered_data[input_columns].copy()
Main_target = filtered_data[target_columns].copy()
```

```python
Main_data
```

```python
import numpy as np
numeric_colms = Main_data.select_dtypes(include = np.number).columns.tolist()
categorical_colms = Main_data.select_dtypes(include = 'object').columns.tolist()
```

```python
# Now we will use the OneHotEncoder to encode the categorical columns.
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output = False, handle_unknown = 'ignore')

encoder.fit(Main_data[categorical_colms])
```

```python
encoder.categories_
```

```python
import pandas as pd

encoder_cols = list(encoder.get_feature_names_out(categorical_colms))

encoder_cols
```

```python
Main_data[encoder_cols] = encoder.transform(Main_data[categorical_colms])
```

```python
Main_data[encoder_cols]
```

```python
numeric_colms, encoder_cols
```

```python
from sklearn.model_selection import train_test_split

splitting_data = Main_data[numeric_colms+encoder_cols].copy()

train_val_dt, test_dt, train_val_target, test_target =  train_test_split(splitting_data, Main_target, random_state=42, test_size = 0.2)

train_data, val_data, train_target, val_target = train_test_split(train_val_dt, train_val_target, test_size=0.25, random_state=42)

```

```python
train_data
```

```python
train_target
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()

model.fit(train_data, train_target)
```

```python
train_predict = model.predict(train_data)
```

```python
from sklearn.metrics import accuracy_score

train_accuracy = accuracy_score(train_target, train_predict)

print("Accuracy of Train_data: ", train_accuracy*100)
```

```python
# Now lets check for the validation data.
```

```python
val_predict = model.predict(val_data)
```

```python
val_accuracy = accuracy_score(val_target, val_predict)

print("Validation_Accuracy : ", val_accuracy*100)


#Our model is performing pretty bad while perforing the model with val_data.
```

```python
# Now we will check min_samples_leaf. by managing this hyperparameter, we can achieve a better model
```

```python
 for leaf in [2,5,10,15,20,22,25,29,30,35,40,45,50,55,60,65,70,75,80]:

  model_3 = DecisionTreeClassifier(criterion = 'gini', splitter='best', min_samples_leaf=leaf)
  model_3.fit(train_data, train_target)

  train_dt_dt_score = model_3.score(train_data, train_target)
  val_dt_dt_score = model_3.score(val_data, val_target)

  print(leaf, train_dt_dt_score, val_dt_dt_score)

  # We have our best model at "min_samples_leaf = 50". at leaf_splitting of 50
```

```python
# NOW LETS CREATE THE BEST MODEL. BY USING ALL THE HYPERPARAMETERS THAT WE FOUND.
```

```python
Best_Model = DecisionTreeClassifier(
    criterion = 'gini',
    splitter = 'best',
    max_depth = 9,
    min_samples_split = 115,
    min_samples_leaf = 50,
    random_state = 42
)

Best_Model.fit(train_data, train_target)

Best_Prediction = Best_Model.predict(train_data)
```
