```python
import opendatasets as od

dataset_url = 'https://www.kaggle.com/jsphyg/weather-dataset-rattle-package'

od.download(dataset_url)
```

```python
import os

data_dir = './weather-dataset-rattle-package'

os.listdir(data_dir)
```

```python
data = data_dir + '/weatherAUS.csv'
```

```python
import pandas as pd

raw_df = pd.read_csv(data)

raw_df
```

```python
#First lets drop the unknown columns in RainToday and RainTommorow columns
```

```python
raw_df.dropna(subset = ['RainToday', 'RainTomorrow'], inplace = True)
```

```python
raw_df.info()
```

```python
#Now let us identify the numeric columns and categorical columns in our whole dataset and we will clear them
```

```python
import numpy as np
#numerical columns

numerical_columns = raw_df.select_dtypes(include = np.number).columns.tolist()
```

```python
numerical_columns
```

```python
#categorical columns

categorical_columns = raw_df.select_dtypes('object').columns.tolist()

categorical_columns
```

```python
# Now lets use imputer technique and replace the null values present in numerical columns
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy = 'mean')

raw_df[numerical_columns] = imputer.fit_transform(raw_df[numerical_columns])
```

```python
#checking if the null values present in the numerical columns

raw_df[numerical_columns].isna().sum()
```

```python
#Now lets purify the categorical columns
```

```python
raw_df[categorical_columns] = raw_df[categorical_columns].fillna('unknown')
```

```python
#checking if the null values presentin categorical columns
```

```python
raw_df[categorical_columns].isna().sum()
```

```python
raw_df.value_counts('RainToday')
```

```python
#Now lets seperate the input_columns and target_columns
```

```python
input_columns = list(raw_df.columns)[1:-1]
```

```python
input_cols = raw_df[input_columns]
target_cols = raw_df['RainTomorrow']
```

```python
#Now our data Set is clear for our Model Implementation. Now we need to scalarize the numerical Columns and Encode the Categorical values to make the best fit for our model
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaler.fit(input_cols[numerical_columns])
```

```python
list(scaler.data_min_)
```

```python
list(scaler.data_max_)
```

```python
input_cols[numerical_columns] = scaler.transform(input_cols[numerical_columns])
```

```python
input_cols[numerical_columns].describe()
```

```python
#We will do the encoding for categorical Columns of inputs_column
```

```python
categorical_columns = input_cols.select_dtypes('object').columns.tolist()
```

```python
input_cols[categorical_columns].info()
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output = False, handle_unknown = 'ignore')

encoder.fit(input_cols[categorical_columns])
```

```python
encoder.categories_
```

```python
#Now we will create names for these new columns using get_feature_names function.
```

```python
encoded_cols = list(encoder.get_feature_names_out(categorical_columns))
```

```python
encoded_cols
```

```python
input_cols[encoded_cols] = encoder.transform(input_cols[categorical_columns])
```

```python
input_cols.columns.to_list()
```

```python
#We have a dataset that contains oldcolumns and new columns. now lets make a perfect dataset for out model.
#by adding the encoded_cols + numerical_columns. because, we need to remove the duplicates
```

```python
Main_input_columns = input_cols[numerical_columns+encoded_cols]
```

```python
Main_input_columns
```

```python
#Now Let Us split the data for training_set, validation_set, test_set and we will design our model
```

```python
from sklearn.model_selection import train_test_split

train_val_input, test_input, train_val_target, test_target = train_test_split(Main_input_columns, target_cols ,test_size=0.2, random_state = 32)
```

```python
#Now Let us split the validation and train inputs

train_input, val_input, train_target, val_target  = train_test_split(train_val_input, train_val_target, test_size = 0.25, random_state=32)
```

```python
print(train_input.shape)
print(val_input.shape)
print(test_input.shape)
print(train_target.shape)
print(val_target.shape)
print(test_target.shape)
```

```python
#Till Here our DATA PREPROCESSING HAS COMPLETED. NOW, LETS TRAIN THE MODEL.
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(train_input, train_target)
```

```python
train_predict = model.predict(train_input)

train_predict
```
